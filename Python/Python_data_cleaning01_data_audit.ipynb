```python
customer_orders = (
    orders.merge(
        customers[["customer_id", "customer_unique_id"]],
        on="customer_id",
        how="left"
    )
)

orders_per_customer = (
    customer_orders.groupby("customer_unique_id")["order_id"]
    .nunique()
)

print("Customers with at least one order:", orders_per_customer.size)
print("Customers with 2+ orders:", (orders_per_customer >= 2).sum())
print("Maximum orders by one customer:", orders_per_customer.max())
```

```python
customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
payments_clean = payments.copy()
reviews_clean = reviews.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
translation_clean = translation.copy()
```

```python
order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in order_date_columns:
    orders_clean[col] = pd.to_datetime(
        orders_clean[col],
        errors="coerce"
    )

reviews_clean["review_creation_date"] = pd.to_datetime(
    reviews_clean["review_creation_date"],
    errors="coerce"
)

reviews_clean["review_answer_timestamp"] = pd.to_datetime(
    reviews_clean["review_answer_timestamp"],
    errors="coerce"
)

order_items_clean["shipping_limit_date"] = pd.to_datetime(
    order_items_clean["shipping_limit_date"],
    errors="coerce"
)
```

```python
products_clean["product_category_name"] = (
    products_clean["product_category_name"]
    .fillna("unknown")
)
```

```python
products_clean = products_clean.merge(
    translation_clean,
    on="product_category_name",
    how="left"
)

products_clean["product_category_name_english"] = (
    products_clean["product_category_name_english"]
    .fillna("unknown")
)
```

```python
# Build the main analytical dataset

order_analysis = (
    order_items_clean
    .merge(
        orders_clean[
            [
                "order_id",
                "customer_id",
                "order_status",
                "order_purchase_timestamp",
                "order_delivered_customer_date",
                "order_estimated_delivery_date"
            ]
        ],
        on="order_id",
        how="left"
    )
    .merge(
        customers_clean[
            [
                "customer_id",
                "customer_unique_id",
                "customer_city",
                "customer_state"
            ]
        ],
        on="customer_id",
        how="left"
    )
    .merge(
        products_clean[
            [
                "product_id",
                "product_category_name_english"
            ]
        ],
        on="product_id",
        how="left"
    )
    .merge(
        sellers_clean[
            [
                "seller_id",
                "seller_city",
                "seller_state"
            ]
        ],
        on="seller_id",
        how="left"
    )
)

print("Rows:", order_analysis.shape[0])
print("Columns:", order_analysis.shape[1])
```
